```python
import numpy as np
import pandas as pd
print("Numpy version",np.__version__)
print("Pandas version",pd.__version__)
```

```python
marks=np.array([78,85,92,67,88])
print("Marks",marks)
print("Type:",type(marks))
```

```python
print("Number of dimensions:",marks.ndim)
print("Shape:",marks.shape)
print("Number of elements:",marks.size)
```

```python
print("Original:",marks)
print("Add 5",marks+5)
print("Multiply by 2",marks*2)
print("Divide by 2:",marks/2)
print("Square:",marks**2)
```

```python
print("Mean:",marks.mean())
print("Sum:",marks.mean())
print("Max:",marks.max())
print("Min:",marks.min())
```

```python
print("First Mark:",marks[0])
print("Sec Mark:",marks[1])
print("Third Mark:",marks[2])
```

```python
print("Index 1 to 3:",marks[1:4])
print("First three:",marks[:3])
print("From index 2:",marks[2:])
print("Even indexes",marks[::2])
```

```python
std_marks=np.array([
    [78,85,92],
    [67,88,76],
    [90,95,89]
])
print(std_marks)
print("Shape:",std_marks.shape)
```

```python
print("Mean of value",std_marks.mean())
print("Column wise mean:",std_marks.mean(axis=0))
print("Row wise mean",std_marks.mean(axis=1))
```

```python
data={
    "Name":["Asha","Bharat","Chetan","Divya","Esha"],
    "Department":["CSE","ISE","CSE","AIML","ISE"],
    "Marks":[72,91,84,95,68],
    "Attendance":[82,94,88,96,75]
}
students=pd.DataFrame(data)
students
```

```python
display(students[["Name","Marks"]])
```

```python
display(students[students["Marks"]>=0])
```

```python
display(students[students["Attendance"]>85])
```

```python
students.iloc[0:3]
```

```python
print("Average:",students["Marks"].mean())
```

```python
print("Max:",students["Marks"].max())
```

```python
print("Students from CSE branch:")
display(students[students["Department"]=="CSE"])
```

```python
print("Students who scored atleast 80 and have attendance above 85%")
display(students[(students["Marks"]>=80) & (students["Attendance"]>85)])
```
